# Local Outlier Factor

```python
from sklearn.neighbors import LocalOutlierFactor
from sklearn.datasets import make_moons

x,y = make_moons(n_samples= 300 , noise = 0.1)
```

```python
x_outliers = np.random.uniform(low  = -3 , high =3 , size = (20, 2))
x = np.vstack((x,x_outliers))
```

```python
out = LocalOutlierFactor(n_neighbors =20 , contamination = 0.1)
y_pred = out.fit_predict(x)
```

```python
y_pred
```

```python
#seperate normal data point>>inliers and outlires

x_inliers = x[y_pred == 1]
x_outliers = x[y_pred == -1]

# plotting
plt.scatter(x_inliers[:,0] , x_inliers[:,1] , color ='b' , edgecolor = 'k' , label = 'normal')
plt.title("local outlier  factor on moons datasets")
plt.scatter(x_outliers[:,0] , x_outliers[:,1] , color ='r' , edgecolor = 'k' , label = 'outlier')
plt.legend()
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller
```

```python
adft =adfuller(stock_data["Close"])
```

```python
adft # null hypothesis of adf test is time serie is non stationay pvalues<0.5 , reject the null hypothesis
#here pvalue > 0.05 it mens series is non ststionary
```

```python
pd.Series(adft[0:4] , index = ["test_statistice" , "p-vlue" , "lag" , " no of data point"])
```

```python
#test for stationarity
def test_stationarity(timeseries):
   #determine the rolling statistics
   rolmean = timeseries.rolling(48).mean() # rolling mean
   rolstd = timeseries.rolling(48).std() # rolling std   #plot rolling statistics
   plt.figure(figsize = (18 , 8))
   plt.grid("both")
   plt.plot(timeseries , color = "blue" , label = "Original", linewidth = 3)
   plt.plot(rolmean , color = "blue" , label = "Rolling mean", linewidth = 3)
   plt.plot(rolstd  , color = "blue" , label = "Rolling_std", linewidth = 3)
   plt.legend(loc = "best" , fontsize = 20 , shadow = True , facecolor = "lightpink")
   plt.title("Rolling mean and Standard devaition " , fontsize = 25)
   plt.xticks(fontsize = 15)
   plt.yticks(fontsize = 15)
   plt.show(block = False)

   print("Result of dickey fuller test")
   adft = adfuller(timeseries = "ATC")
   #out for dft will give us without defining what the values are
   #hences we are manullaly write what values does it explain using a for loop
   output = pd.Series(adft[0:4] , index = ["Test Statistics" , "p-value" , "no. of lags used " , "Number of"])
   for key , values in adft[4].items():
      output['criticle value (%s)' %key] = values
   print(output)
```

```python
test_stationarity(stock_data.Close)
```
