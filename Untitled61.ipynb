# Kmeans clustering Hierarchical clustering

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn import datasets
data = datasets.load_iris()
x = data.data
y = data.target
```

```python
x
```

```python
y
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler
```

```python
x_scaled = scaler.fit_transform(x)
x_scaled
```

```python
#appling pca
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
pca_scaled = pca.fit_transform(x_scaled)
```

```python
pca_scaled
```

```python
plt.scatter(pca_scaled[: , 0] , pca_scaled[: , 1] , c = y)
```

```python
#aggolomarative clustering >> make dendogram

import scipy.cluster.hierarchy as sc

plt.figure(figsize = (20 , 7))
plt.figure("dendrogram")

sc.dendrogram(sc.linkage(pca_scaled , method = "ward"))
plt.title("dendrogram")
plt.xlabel("sample data")
plt.ylabel("Euclidian Distance")
```

```python
from sklearn.cluster import AgglomerativeClustering
cluster= AgglomerativeClustering(n_clusters = 3 , affinity = "euclidean" , linkage = "ward")
cluster.fit(pca_scaled)
```

```python
cluster.labels_
```

```python
plt.scatter(pca_scaled[: , 0] , pca_scaled[: , 1] , c = cluster.labels_)
```

```python
#k start from 2
from sklearn.metrics import silhouette_score
silhouette_coeff = []
for k in range(2 , 11):
    agglo = AgglomerativeClustering(n_clusters= k , affinity = "euclidean" , linkage = "ward")
    agglo.fit(x_scaled)
    score = silhouette_score(x_scaled , agglo.labels_)
    silhouette_coeff.append(score)
```

```python
silhouette_coeff
```

```python
# plotting silhouette score
plt.plot(range(2,11), silhouette_coeff)
plt.xlabel("number of cluster")
plt.ylabel("silhouette coeff")
plt.show()
```

# DBSCAN

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons

x,y = make_moons(n_samples = 250 , noise = 0.10)
```

```python
x
```

```python
plt.scatter(x[:, 0] , x[:, 1] , c = y)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
x_scaled
```

```python
dbscan = DBSCAN(eps = 0.3)
```

```python
dbscan
```

```python
dbscan.fit(x_scaled)
```

```python
dbscan.labels_
```

```python
plt.scatter(x[:, 0] , x[:, 1] , c = dbscan.labels_)
```
